### Завдання 25

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]], dtype=float)
means = np.mean(A, axis=0)
stds = np.std(A, axis=0)
Z = (A - means) / stds
print("Середні значення стовпців:", means)
print("Стандартні відхилення:", stds)
print("Нормалізована матриця:\n", Z)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
A = np.array([[1, 2, 1],
              [2, 1, -1],
              [3, -1, 2]])
b = np.array([9, 8, 3])
solution = np.linalg.solve(A, b)
calculated_b = A @ solution
print("Розв'язок [x, y, z]:", solution)
print("A @ x:", calculated_b)
print("Вектор b:", b)
print("Результати збігаються:", np.allclose(calculated_b, b))
````
